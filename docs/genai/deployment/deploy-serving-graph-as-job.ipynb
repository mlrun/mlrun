# Run a serving graph as a batch job
You can convert a ServingRuntime to a KubejobRuntime, so that the graph can be run as a standalone job. This is a simpler implementation that can be run on th fly, and runs much faster than the same function as a serving runtime. YOu can use all job parameters that are used in serving graphs. Any resources assigned to the serving function are assigned to the new job. The original serving function remains unchanged and can still be invoked after running the job. 
You can run the resulting Kubejob function from the UI after `to_job()` was performed.

To run an existing graph as a job, just replace: 
`function.deploy()` 
qith 
`function.to_job()`

{py:meth}~`mlrun.runtimes.ServingRuntime.to_job` has an optional parameter `func_name` to rename the job function. If you do not specify a name, MLRun automatically appends the suffix '-batch' to the serving function name to differentiate between the functions.


Limitations:
- Scheduling of this function is not supported. 
- Serving graph as a job with child functions is not supported.
