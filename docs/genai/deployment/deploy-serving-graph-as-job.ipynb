# Run a serving graph as a batch job
You can convert a ServingRuntime to a KubejobRuntime, so that the graph can be run as a standalone job.

Use the {py:meth}~`mlrun.runtimes.ServingRuntime.to_job` method to convert a ServingRuntime. Mlrun generates a new KubejobRuntime based on the serving function. `to_job` has an optional parameter `func_name` to rename the job function. If you do not specify a name, MLRun automatically appends the suffix '-batch' to the serving function name to prevent database collision. The original serving function remains unchanged and can still be invoked after running the job.


Guidelines:
- Any resources assigned to the serving function are assigned to the new job.
- Scheduling of this function is not supported. 
- Serving graph as a job with child functions is not supported.
- You can run the resulting Kubejob function from the UI after `to_job()` was performed.
