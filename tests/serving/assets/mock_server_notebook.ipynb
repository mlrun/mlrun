```python
import mlrun
from typing import List


def _handle_post_process(_data):
    return [b + 1 for b in _data]


def _handle_pre_process(_data):
    return _data * 2
```

```python
class ClassifierModel(mlrun.serving.V2ModelServer):
    def load(self):
        pass

    def predict(self, body: dict) -> List:
        input_length = len(body["inputs"])
        return list(range(input_length))


class PreClassifierModel(ClassifierModel):
    def preprocess(self, body, op):
        if op in ["predict", "infer"]:
            body["inputs"] = _handle_pre_process(body["inputs"])
        return body


class PostClassifierModel(ClassifierModel):
    def postprocess(self, body):
        body["outputs"] = _handle_post_process(body["outputs"])
        return body
```
